```python
#object oriented API

import matplotlib.pyplot as plt
import pandas as pd

data = {
    "sal" :[20000,34000,66000,86000,70000,56000,89000,67000,67000,77000,88000],
    "age" :[21,23,35,46,47,29,46,44,41,42,60],
    "dept":["HR","IT","fiance","Teacher","fiance","IT","HR","Teacher","IT","HR","IT"]
}

d = pd.DataFrame(data)
d
```

```python
fig,axs = plt.subplots(1,3,figsize = (10,5))
```

```python
#lineplot
sort_age = d.sort_values("age")
```

```python
sort_age
```

```python
fig,axs = plt.subplots(1,3,figsize = (13,5))





# line plot
axs[0].plot(sort_age["age"],d["sal"],color="red",marker = "*",linewidth = "2",markersize = 2)
axs[0].grid()
axs[0].set_title("Line plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

#histogram
axs[1].hist(d["sal"],bins =5,color = "skyblue")
axs[1].set_title("histogram")
axs[1].set_xlabel("Sal")
axs[1].set_ylabel("Frequency")



# boxplot
axs[2].boxplot(d["sal"])
axs[2].set_title("Boxplot")
axs[2].set_xlabel("sal")

plt.savefig("multipleplots.png")
plt.show()
```
