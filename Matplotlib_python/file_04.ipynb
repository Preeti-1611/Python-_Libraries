```python
#Bivariate : numerical - categorical
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Dept":["HR","fiance","Tec","IT","HR","IT","fiance","HR","HR"],
    "Sal":[23000,34000,56000,46000,20000,89000,45000,75000,67000]
}
```

```python
d = pd.DataFrame(data)
```

```python
d
```

```python
hr_sal = d[d["Dept"]=="HR"]["Sal"]
hr_sal
```

```python
fin_sal = d[d["Dept"] == "fiance"]["Sal"]
fin_sal
```

```python
it_sal = d[d["Dept"]=="IT"]["Sal"]
it_sal
```

```python
#boxplot

plt.boxplot([hr_sal,fin_sal,it_sal],labels =["HR","Fin","IT"])
plt.grid()
plt.show()
```

```python
#pie chart
```

```python
salary_by_dept = d.groupby("Dept")["Sal"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept,labels = salary_by_dept.index ,autopct = "%1.2f")
plt.axis("equal")
plt.show()
```

```python
#bar plot

hr_mean = sum(hr_sal)/len(hr_sal)
fin_mean = sum(fin_sal)/len(fin_sal)
it_mean = sum(it_sal)/len(it_sal)
```

```python
plt.bar(["HR","fin","IT"],[hr_mean,it_mean,fin_mean],color=["green","red","black"])
plt.show()
```
