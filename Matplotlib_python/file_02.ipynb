```python
#univairiate Numerical
# Univariate analysis is used to understand and summarize one variable at a time. It helps us check distribution, spread, outliers, missing values, and basic statistics of a feature.
# It is the first and most important step 
# in data cleaning and exploratory data analysis.
import matplotlib.pyplot as plt
import pandas as pd
```

```python

data = {
    "Sal": [29000,34000,45000,24000,45000,46000,34000,42000,90000,44000,37000,33000]
}
```

```python
d = pd.DataFrame(data)
```

```python
d
```

```python
d.head()
```

```python
# A line plot is a type of graph that displays data points connected by
# a line. It is used to visualize trends over a continuous period of time, 
# such as days, months, or years. Line plots help identify patterns, 
# increases, decreases, or fluctuations in data.


#lineplot
plt.plot(d["Sal"],color = "red",marker ="*",linestyle ="--",linewidth = "2")
plt.grid()
plt.show()
```

```python
#Histogram
# A histogram shows how many values fall into a range (bin).

# 👉 It is not about individual values
# 👉 It is about groups/ranges of values


# A histogram is a graphical representation of the distribution 
# of numerical data. It groups data into bins (ranges) and shows how many 
# values fall into each bin. It helps identify 
# the shape of the data distribution, such as normal, skewed, or uniform.

plt.hist(d["Sal"])
plt.show()
```

```python
plt.hist(d["Sal"],bins =7,color = "green")
plt.show()
```

```python
#boxplot
plt.boxplot(d["Sal"])
plt.grid()
plt.show()
```

```python
# Skewness describes how data is distributed. 
# If the data is not symmetrical and one tail is longer,
# it is called skewed. In a right-skewed distribution, 
# the tail is on the right and most values are low. In 
# a left-skewed distribution
# , the tail is on the left and most values are high.
# # histogram


import matplotlib.pyplot as plt

marks = [10,20,25,30,35,40,50,60,70,80]

plt.hist(marks, bins=5)
plt.xlabel("Marks Range")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
labels = ['Mango', 'Apple', 'Banana', 'Grapes']
sizes = [40, 25, 20, 15]

# Pie Chart
plt.pie(sizes, labels=labels,autopct ="1.1f%%")

# Title
plt.title("Students' Favourite Fruits")

# Show the chart
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ['Mango', 'Apple', 'Banana', 'Grapes']
sizes = [40, 25, 20, 15]

explode = [0.1, 0, 0, 0]   # Highlight Mango slice

plt.pie(sizes, labels=labels, explode=explode, shadow=True, autopct="%1.1f%%")
plt.title("Favourite Fruits")
plt.show()
```
