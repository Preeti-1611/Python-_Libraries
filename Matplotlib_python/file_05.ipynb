```python
import matplotlib.pyplot as  plt
import pandas as pd
#Multivariate analysis 3 numerical
```

```python
data = {
    "sal" :[59000,34000,66000,86000,20000,78000,89000,67000,67000,77000,88000],
    "age" :[34,23,35,46,21,33,46,44,41,42,30],
    "dept":["HR","IT","fiance","Teacher","fiance","IT","HR","Teacher","IT","HR","IT"]
}
```

```python
d = pd.DataFrame(data)
```

```python
d["exp"] = [4,1,5,8,2,4,7,6,5,3,2]
```

```python
#bubbleplot:
plt.scatter(d["age"],d["sal"],s = d["exp"]*48,color ="blue",edgecolor ="black")
plt.show()

```

```python
import matplotlib.pyplot as plt

x = [5, 2, 8]           # Population
y = [400, 200, 800]     # Area
size = [70, 40, 90]     # Pollution index

plt.scatter(x, y, s=size)

plt.xlabel("Population (millions)")
plt.ylabel("Area (sq km)")
plt.title("Bubble Plot Example")
plt.show()
```
