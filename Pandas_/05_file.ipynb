```python
#data cleaning
import numpy as np
import pandas as pd

d1 = {
    "Name":["preeti","ram","Ravi","kavi","shree"],
    "Age":[23,43,26,57,np.nan],
    "Dept":["Hr","Teacher","Fiance",np.nan,"Hr"],
    "Sal":[34000,45000,np.nan,450000,34000]
}
```

```python
d = pd.DataFrame(d1)
```

```python
d
```

```python
d.isnull().sum()
```

```python
#remove the all null values
d.dropna()
```

```python
d
```

```python
d.dropna(how = "any") #any row that had null values
```

```python
d.dropna(how = "all") #if all the values in any row are null then drop
```

```python
#missiing values getting rid - dropna(),filling themup - fillna()

d.fillna(0)
```

```python
d["Age"].fillna(d["Age"].mean())
```

```python
#forward fill backward fill


d["Age"].fillna(method = "ffill")
```

```python
d["Sal"].fillna(method = "bfill")
```

```python
d["Name"].replace("preeti","murali")
```

```python
d["Name"]=d["Name"].replace("preeti","murali")
```
