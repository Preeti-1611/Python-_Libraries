```python
#conditional selection
import pandas as pd
fruits = {
    'apple':0.3,
    'Banana':3.4,
    'Mango':1.2,
    'kiwi':5.0,
    'cherries':2.7,
    'orange':6.0,
    'painaple':8.0
}
```

```python
s = pd.Series(fruits,name = 'protien')
```

```python
s
```

```python
s>1
```

```python
s[s>2]
```

```python
#logical operate = and ,or, not

(s>0.5)& (s>2)



```

```python
s[(s>0.5) & (s>2)]
```

```python
s[(s>0.5) | (s>2)]
```

```python
s[~(s>1)]
```
