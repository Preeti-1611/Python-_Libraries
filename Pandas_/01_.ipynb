```python
s = pd.Series([10, 20, 30], name="MySeries")
print(s)
# Output shows: Name: MySeries, dtype: int64
```

```python
s2 = s.rename("NewName")
print(s2.name)  # → "NewName"
```

```python
s = pd.Series([2.0, 3.1, 4.5], index=['a', 'b', 'c'])
print(s.loc['b'])         # → 3.1
print(s.loc['a':'c'])     # range from 'a' to 'c' **includes** 'c' :contentReference[oaicite:3]{index=3}
```

```python
print(s.loc[['a', 'c']])
```

```python
s = pd.Series([2.0, 3.1, 4.5], index=['a', 'b', 'c'])
print(s.iloc[1])        # → 3.1 (because it's the second element)
print(s.iloc[0:2])      # slice first 2 by position :contentReference[oaicite:5]{index=5}
```

```python
# dtype: Tells the data type of the Series values.

# index: The labels for the Series; can be custom or default.

# Rename: Use Series.rename() to change the name or index labels.

# Indexing:

# Use s[label] for direct label-based access.

# Use slicing s[start:stop] (but be careful with whether it's positional or label-based).

# .loc: For label-based selection (works with index names).

# .iloc: For integer position-based selection.
```
