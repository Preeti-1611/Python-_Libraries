```python
# Definition:
# Take a dataset of employee names and salaries (manually entered). Sort the data based on salary without using sorted() or sort() — use any manual sorting logic (bubble sort, selection sort).
# Concepts: Nested loops, tuples/lists, sorting logic.
```

```python
emp = [
    ("Preeti", 50000),
    ("Ravi", 30000),
    ("Anu", 45000),
    ("Kiran", 60000),
    ("Manu", 25000)
]

n = len(emp)

for i in range(n):
    min_index = i
    for j in range(i + 1, n):
        if emp[j][1] < emp[min_index][1]: 
            min_index = j

    emp[i], emp[min_index] = emp[min_index], emp[i]  

print("Sorted by Salary :")
for emp1 in emp:
    print(emp1)
```
