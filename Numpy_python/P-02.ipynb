```python
matrix = np.array([
    [1, 2, 3],   # row 0
    [4, 5, 6],   # row 1
    [7, 8, 9]    # row 2
])
```

```python
matrix[0]
```

```python
matrix[2]
```

```python
matrix[1]
```

```python
# Selecting a COLUMN
#Use : for all rows.
matrix[:,2]
```

```python
matrix[:, 0]   # column 0
```

```python
#  Selecting a SINGLE ELEMENT

# Format: [row, column]

matrix[1, 2]

```

```python
matrix[2,1]
```

```python
# Slicing part of the matrix (sub-matrix)
# ✔ Example: First 2 rows
matrix[0:2, :]
```

```python
# ✔ Example: First 2 columns
matrix[:, 0:2]
```

```python
#  5. Middle block (5,6,8,9)
matrix[1:3, 1:3]
```
