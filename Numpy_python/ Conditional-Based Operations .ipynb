```python
import numpy as np
arr = np.array([10, 60, 30, 80])

# Get positions where arr > 50
print(np.where(arr > 50))      # (array([1, 3]),)

# Apply if-else
print(np.where(arr > 50, "Pass", "Fail"))
# ['Fail' 'Pass' 'Fail' 'Pass']
```
